```python
from transformers import pipeline
Voice_to_text=pipeline('automatic-speech-recognition',model="openai/whisper-base")
Vts=Voice_to_text('output_audio.mp3')
print(Vts)
```

```python
from google.colab import files
uploaded=files.upload()
```

```python
text_generator=pipeline('text-generation')
generate_Text=[]
for i in range(10):
  text = text_generator(Vts['text'])
  generate_Text.append(text[0])
for idx, txt in enumerate(generate_Text):
    print(f"Generated Text {idx}: {txt}")
```

```python
from transformers import pipeline

Sentiment_analyzer = pipeline("sentiment-analysis")


sentiment_result = []

for i in generate_Text:
    sentiment = Sentiment_analyzer(i['generated_text'])[0]
    sentiment_result.append(sentiment)

for lists in sentiment_result:
  print(lists)

top_priority = sentiment_result[0]

for i in sentiment_result:
    if i['score'] > top_priority['score']:
      top_priority = i

print("top_priority" , top_priority)
```
